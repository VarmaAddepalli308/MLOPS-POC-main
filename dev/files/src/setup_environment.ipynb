```python
# One-off / idempotent bootstrap: catalog and schemas for the live-data project.
import sys
import os
sys.path.insert(0, os.getcwd())

from utils import load_project

cfg = load_project()
dbutils.widgets.text("catalog", "stock_demo")
catalog = dbutils.widgets.get("catalog")

try:
    spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
except Exception as e:
    print(f"Could not create catalog {catalog} ({type(e).__name__}); assuming it already exists.")

for schema in cfg["schemas"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")

print(f"Environment ready for catalog '{catalog}'. Yahoo Finance will bootstrap historical OHLCV directly into Delta.")
```
